### Check de descarga automática

```python
def check_raw_dades_obertes(response: requests.Response) -> None:
    """
    Levanta un error si los datos no tienen el formato esperado
    de la API de Dades Obertes Catalunya.
    """
    mensaje_error = "Error de formato para los datos de la API de Dades Obertes Catalunya."

    response.raise_for_status()
    data = response.json()

    # response.json() debe ser una lista
    if not isinstance(data, list) or not data:
        raise ValueError(
            f"{mensaje_error}\n"
            "response.json() no es una lista."
            f"URL: {url}"
        )

    for row in data: # Asumimos que cada elemento de la lista es una fila

        # Cada fila debe ser un diccionario
        if not isinstance(row, dict) or set(row) != {"count", "any"}:
            raise ValueError(
                f"{mensaje_error}\n"
                "Una fila de la lista no es un diccionario."
                f"URL: {url}"
            )
```

## Limpieza genérica

```python
def filtrar_enteros(df: pd.DataFrame, columna: str):
    """
    Dado un DataFrame y una columna, modifica el DataFrame para
    conservar solamente las filas en las que esta columna tiene un número entero (en cualquier formato).
    La columna de entrada pasará a tener formato int.
    """

    # Comprobar que el DataFrame contiene la columna
    if columna not in df.columns:
        raise ValueError(f"La columna '{columna}' no existe en el DataFrame")

    # Convertir a numérico; los valores no válidos pasan a NaN
    valores = pd.to_numeric(df[columna], errors="coerce")

    # Conservar únicamente los valores enteros
    mascara = valores.notna() & (valores % 1 == 0)

    # Filtrar y convertir la columna a int
    df.drop(df.index[~mascara], inplace=True)
    df[columna] = valores.loc[mascara].astype(int)

    return df
```

## Checks genéricos

```python
def comprobar_esquema(
    df: pd.DataFrame,
    columnas: Optional[list] = None,
    filas: Optional[list] = None,
    modo_columnas: str = "contiene",
    modo_filas: str = "contiene"
) -> None:
    """
    Comprueba que un DataFrame contiene las filas y columnas indicadas.

    Args:
        df: DataFrame a comprobar.
        filas: índices esperados (None para no comprobar).
        columnas: columnas esperadas (None para no comprobar).
        modo_filas: modo de comprobación de filas
        modo_columnas: modo de comprobación de columnas

        Ambos modos pueden ser:
            - "contiene": permite columnas/filas adicionales.
            - "coincide": exige las mismas columnas/filas, sin importar el orden.
            - "estricto": exige las mismas columnas/filas y en el mismo orden.

    Raises:
        ValueError: si el DataFrame no cumple alguna condición
                    o si se indica un modo no válido.
        SyntaxError: si no se pasan filas ni columnas, 
                    pero se piden modos "coincide" o "estricto" en ambas dimensiones.
    """

    modos_validos = {"contiene", "coincide", "estricto"}

    # Comprobar modos
    if modo_filas not in modos_validos:
        raise ValueError(f'Modo de filas no válido: {modo_filas}. El modo puede ser "contiene", "coincide" o "estricto".')

    if modo_columnas not in modos_validos:
        raise ValueError(f'Modo de columnas no válido: {modo_columnas}. El modo puede ser "contiene", "coincide" o "estricto".')

    # Caso sin sentido
    if all((modo_columnas != "contiene", columnas is None, modo_filas != "contiene", filas is None)):
        raise SyntaxError('No tiene sentido pedir que un DataFrame esté vacío.')

    # Comprobar columnas
    if columnas is not None or modo_columnas != "contiene":
        esperadas = pd.Index(columnas if columnas is not None else [])

        if modo_columnas == "estricto":
            if not df.columns.equals(esperadas):
                raise ValueError("Las columnas no coinciden exactamente o no están ordenadas.")

        elif modo_columnas == "coincide":
            if len(df.columns) != len(esperadas) or not esperadas.isin(df.columns).all():
                raise ValueError("Las columnas no coinciden exactamente.")

        elif modo_columnas == "contiene":
            if not esperadas.isin(df.columns).all():
                raise ValueError("Faltan columnas en el DataFrame.")
            
    # Comprobar filas
    if filas is not None or modo_filas != "contiene":
        esperadas = pd.Index(filas if filas is not None else [])

        if modo_filas == "estricto":
            if not df.index.equals(esperadas):
                raise ValueError("Las filas no coinciden exactamente o no están ordenadas.")

        elif modo_filas == "coincide":
            if len(df.index) != len(esperadas) or not esperadas.isin(df.index).all():
                raise ValueError("Las filas no coinciden exactamente.")

        elif modo_filas == "contiene":
            if not esperadas.isin(df.index).all():
                raise ValueError("Faltan filas en el DataFrame.")
```

```python
def check_data(data: list):
    "Comprueba que los datos están en el formato esperado."

    for row in data:
        if not isinstance(row, dict) or set(row) != {"count", "any"}:
            raise ValueError("Each row must contain exactly 'count' and 'any'.")

        if not isinstance(row["count"], str) or not row["count"].isdigit():
            raise ValueError("'count' must be a string of digits.")

        if not isinstance(row["any"], str) or not (
            row["any"].isdigit() or
            (row["any"][:4].isdigit() and row["any"][4:] == "_1S")
        ):
            raise ValueError("'any' must be a year or a year with '_1S'.")

    years = [row["any"] for row in data]
    if len(years) != len(set(years)):
        raise ValueError("'any' values must be unique.")


def filter_data(data: list) -> list:
    "Erases years not in format YYYY."
    return [row for row in data if isinstance(row, dict)
            and isinstance(row.get("any"), str)
            and len(row["any"]) == 4
            #and row["any"].isdigit()
            ]
```
