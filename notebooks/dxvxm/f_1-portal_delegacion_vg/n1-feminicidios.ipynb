```python
df_raw = pd.read_csv(PATH_INPUT / "f_1_1-feminicidios_raw.csv")
df_raw.info()
df_raw.head(10)
```

Con el formato indicado para la descarga, tenemos columnas de 2 tipos:

- Columnas de variables:
    - Comunidad autónoma (As)
    - Provincia (As)
    - Año
    - Denuncia -2-

- Columna de indicador:
    - Feminicidios pareja o expareja
    - Huérfanas y huérfanos menores de edad -1-

Este formato facilita:

- Filtrar según las variables fácilmente.
- Coger la medida objetivo fácilmente.

Además, cuando ambos indicadores son 0, la fila desaparece, lo que reduce el número de valores (con esto hay que tener cuidado).

```python
# Cambiar el nombre de las columnas estándar
df_standard = df_raw.copy().rename(columns={
    "Comunidad autónoma (As)": CCAA,
    "Provincia (As)": PROVINCIA,
    "Año": ANYO
})
df_standard.head(1)
```

```python
# Por ejemplo:
    # ¿Cuántos feminicidios (indicador) hubo en la Provincia de Barcelona (filtro 1) en 2023 (filtro 2)?

total = df_standard.loc[
    (df_standard[PROVINCIA] == "Barcelona") &             # Filtro 1
    (df_standard[ANYO] == 2023),                                # Filtro 2
    "Feminicidios pareja o expareja"                        # Indicador
].sum()

print(f"Feminicidios en Barcelona en 2023: {int(total)}")
```

### dxvxm-espanya-v_2_1_1-esp

Este es el pie chart. Con filtrar los años y agrupar, sirve (a no ser que nos interese desglosar por años).

```python
# Eliminar todas las columnas excepto denuncia y el indicador
df_pie = df_clean_2008[["Denuncia -2-", "Feminicidios pareja o expareja"]]

# Agrupar por denuncia
df_pie = df_pie.groupby("Denuncia -2-").sum().reset_index()

# Renombrar columna
df_pie.rename(columns={
    "Feminicidios pareja o expareja": f"2008-{ANYO_OBJETIVO}"
}, inplace=True)

# Renombrar "filas"
df_pie["Denuncia -2-"] = df_pie["Denuncia -2-"].replace({
    "No": "No había denuncia",
    "Sí": "Había denuncia"
})

df_pie.head()
```

```python
df_pie.to_csv(PATH_OUTPUT / "v_2_1_1-esp.csv", index=False)
```

### dxvxm-espanya-v_2_1_2-esp

No se puede hacer que el Flourish use una columna para discriminar entre gráficos solapados.

```python
# Eliminar todas las columnas excepto, año, denuncia y el indicador
df_combo = df_clean_2008[[ANYO, "Denuncia -2-", "Feminicidios pareja o expareja"]]

# Agrupar por denuncia y año
df_combo = df_combo.groupby([ANYO, "Denuncia -2-"]).sum().reset_index()

# Por un lado, buscar los totales
df_totales = df_combo.drop(columns="Denuncia -2-").groupby([ANYO]).sum().reset_index() \
    .rename(columns={"Feminicidios pareja o expareja": "Total víctimas"})

# Por otro, buscar los casos con denuncia
df_denuncia = df_combo[df_combo["Denuncia -2-"] == "Sí"] \
    .drop(columns="Denuncia -2-").groupby([ANYO]).sum().reset_index() \
    .rename(columns={"Feminicidios pareja o expareja": "Había denuncia"})

# Hacer el join con los años
df_combo = df_totales.merge(df_denuncia, on=ANYO)

df_combo.head(20)
```

```python
df_combo.to_csv(PATH_OUTPUT / "v_2_1_2-esp.csv", index=False)
```

### dxvxm-espanya-v_6_3_2-esp

En Flourish, para que diversas series aparezcan juntas, no sirve el row filter (esto serviría para que el usuario seleccione una, pero no le dejaŕia verlas todas).

```python
# Filtrar años
df_huerfanos_anyo_ccaa = df_clean[df_clean[ANYO] >= 2013]

# Eliminar todas las columnas excepto año, CCAA y el indicador
df_huerfanos_anyo_ccaa = df_huerfanos_anyo_ccaa[[ANYO, CCAA, "Huérfanas y huérfanos menores de edad -1-"]]

# Agrupar por año y CCAA
df_huerfanos_anyo_ccaa = df_huerfanos_anyo_ccaa.groupby([ANYO, CCAA]).sum().reset_index()

df_huerfanos_anyo_ccaa.head()
```

```python
# Pivotar  año, CCAA y el indicador
df_huerfanos_anyo_ccaa_wide = df_huerfanos_anyo_ccaa.pivot(
    index="Año",
    columns="CCAA",
    values="Huérfanas y huérfanos menores de edad -1-"
).reset_index().rename_axis(None, axis=1).fillna(0)

# Pasar todas las columnas a enteros
df_huerfanos_anyo_ccaa_wide = df_huerfanos_anyo_ccaa_wide.astype({
    col: int
    for col in df_huerfanos_anyo_ccaa_wide.columns
})

# Añadir la media de todas las columnas excepto año
df_huerfanos_anyo_ccaa_wide["Media España"] = df_huerfanos_anyo_ccaa_wide.drop(columns=["Año"]).mean(axis=1)

df_huerfanos_anyo_ccaa_wide.head(5)
```

### dxvxm-lleida-v_1_2

```python
df_clean.head()
```

```python
indicador = "Feminicidios pareja o expareja"

# Seleccionar columnas
df = df_clean[[CCAA, PROVINCIA, ANYO, indicador]]

# 3 selecciones + filtros + agrupaciones:

    # 1. Agrupar por año para tener el total de España
df_es = df[[ANYO, indicador]] \
    .groupby(ANYO).sum().reset_index() \
    .rename(columns={indicador: "España"})

    # 2. Filtrar CCAA y agrupar por año para tener el total de la CCAA
df_cat = df[[ANYO, CCAA, indicador]] \
    [df[CCAA] == "Cataluña"].drop(columns=[CCAA]) \
    .groupby(ANYO).sum().reset_index() \
    .rename(columns={indicador: "Cataluña"})

    # 3. Filtrar por provincia y agrupar por año para tener el total de la provincia
df_lleida = df[[ANYO, PROVINCIA, indicador]] \
    [df[PROVINCIA] == "Lleida"].drop(columns=[PROVINCIA]) \
    .groupby(ANYO).sum().reset_index() \
    .rename(columns={indicador: "Provincia de Lleida"})

# Juntar los tres datasets a partir del año, llenando los NaNs con ceros
df_final = (
    df_lleida
    .merge(df_cat, on=ANYO, how="outer")
    .merge(df_es, on=ANYO, how="outer")
    .fillna(0)
)

# Pasar todas las columnas a enteros
df_final = df_final.astype({
    col: int
    for col in df_final.columns
})

# Ordenar por año
df_lleida = df_final.sort_values(ANYO)

df_lleida.head()
```

```python
# Exportar a csv

# 1. esp
df_lleida.to_csv(PATH_OUTPUT / "lleida-v_1_2-esp.csv", index=False)

# 2. cat
df_lleida_cat = df_lleida.rename(columns={
    'Provincia de Lleida': 'Província de Lleida',
    'Cataluña' : 'Catalunya',
    'España' : 'Espanya'
})   
df_lleida_cat.to_csv(PATH_OUTPUT / "lleida-v_1_2-cat.csv", index=False)
```
