Seguramente no nos interese un nivel por debajo del municipio.

Lo ideal sería tener todas las desagregaciones. Hay varias formas de modelar esto.

- Usando un join:
    - Una tabla con la población por municipio, sexo y edad.

    - Otra tabla que mapee municipios con sus comarcas y provincias (esta, idealmente, sería estándar para todo IBOD)

    - Agregar supone hacer un join.

- Usando una misma tabla:
    - Todas las filas tendrían que tener "municipio", "comarca" y "provincia".

    - Es lo mismo pero ahorrándote el join. Ocupa más espacio. Parece mejor opción que la anterior.

- Usando tablas separadas.
    - Agregar es simplemente coger una tabla diferente. Es como lo ofrece Idescat en este caso.
