Este formato es diferente al del dataset 1.

Poner las CCAA en las columnas permite tener directamente la tabla que nos interesa subir al Florusish.

Tan solo hay que:
- Sustituir los NaNs por zeros, ya que en la interfaz web no hay NaNs, pero sí hay zeros (al revés que en la descarga).
    - Hay un cierto riesgo: puede ser que en un futuro aparezcan NaNs y los codifiquemos como zero.
- Poner los nombres de las columnas en el idioma que toque para la visualización.

## Limpieza

```python
# Convertir NaNs en zeros
d2 = d2_raw.fillna(0)

# Comprobamos que todos los números son enteros
columnas_numericas = d2.select_dtypes(include='number').columns

assert (d2[columnas_numericas] % 1 == 0).all().all(), "Error de formato. Un valor que debía ser un entero no se ha podido convertir a entero."

# Pasamos los valores de las columnas numéricas a enteros
columnas_numericas = d2.select_dtypes(include='number').columns
d2[columnas_numericas] = d2[columnas_numericas].astype(int)

d2.dtypes
```
