## Descarga automática por API

```python
url_data = f"https://analisi.transparenciacatalunya.cat/resource/{resource_code}.json"
```

```python
query = """
SELECT
    any AS Any,
    edat,
    sum(v_ctimes_ateses) AS Víctimes,
    sum(agressors) AS Agressors
GROUP BY
    any,
    edat
ORDER BY
    any,
    edat
"""

response = requests.get(
    url_data,
    params={"$query": query},
    timeout=60
)
check_raw_dades_obertes(response)

df_victimas_agresores_por_edad = pd.DataFrame(response.json())

print(df_victimas_agresores_por_edad.info())
df_victimas_agresores_por_edad.head(20)
```
