Esta puede ser una forma más estándar de publicar los datos, que no poniendo cada tipo de delito en una columna.

```python
# Generamos los años de interés
years = "(" + ", ".join(f"'{year}'" for year in range(2022, 2026)) + ")"
```

```python
categorias = [
    "Delictes contra el patrimoni i contra l’ordre socioeconòmic",
    "De les lesions",
    "Delictes contra la llibertat",
    "Delictes contra la seguretat viària"
]

territorios = {
    "Catalunya": None,
    "ABP* Santa Coloma de Gramenet": "ABP Santa Coloma de Gramenet",
    "ABP* Badalona": "ABP Badalona"
}

def generar_consulta(years, abp=None):

    filtro_abp = ""

    if abp is not None:
        filtro_abp = f"AND rea_b_sica_policial_abp = '{abp}'"

    return f"""
    SELECT
        any,
        t_tol_codi_penal,
        SUM(coneguts) AS total
    WHERE
        any IN {years}
        {filtro_abp}
    GROUP BY
        any,
        t_tol_codi_penal
    ORDER BY
        any,
        t_tol_codi_penal
    """

def obtener_datos(years, abp=None):

    query = generar_consulta(years, abp)

    datos = query_dades_obertes(resource_code, query)

    datos["any"] = pd.to_numeric(
        datos["any"],
        errors="raise"
    ).astype(int)

    datos["total"] = pd.to_numeric(
        datos["total"],
        errors="coerce"
    ).fillna(0)

    return datos

def generar_tabla(datos, etiqueta):

    # Agrupación por año y categoría
    agrupado = (
        datos.groupby(["any", "t_tol_codi_penal"])["total"]
        .sum()
        .unstack(fill_value=0)
    )

    # Garantizamos la presencia de todas las categorías (?)
    resultado = agrupado.reindex(columns=categorias, fill_value=0)

    # Calculamos las categorías restantes
    otras = agrupado.drop(columns=categorias, errors="ignore")

    #resultado["Altres"] = otras.sum(axis=1)

    # Recuperamos el año como columna
    resultado = resultado.reset_index()

    # Añadimos la etiqueta territorial
    resultado.insert(1, "Etiqueta Fila", etiqueta)

    return resultado

tablas = []
for etiqueta, abp in territorios.items():
    tablas.append(generar_tabla(obtener_datos(years, abp), etiqueta))
tabla_final = pd.concat(tablas, ignore_index=True)

orden = {
    etiqueta: i
    for i, etiqueta in enumerate(territorios)
}

tabla_final["_orden"] = (tabla_final["Etiqueta Fila"].map(orden))

tabla_final = (
    tabla_final
    .sort_values(["_orden", "any"])
    .drop(columns="_orden")
    .rename(columns={"any": "Any"})
    .reset_index(drop=True)
)

tabla_final.columns.name = None

display(tabla_final)
```

```python
tabla_final_esp = tabla_final.rename(columns={
    "Delictes contra el patrimoni i contra l’ordre socioeconòmic": "Delitos contra el patrimonio y contra el orden socioeconómico",
    "De les lesions": "De las lesiones",
    "Delictes contra la llibertat": "Delitos contra la libertad",
    "Delictes contra la seguretat viària": "Delitos contra la seguridad vial",
    "Altres": "Otros"
})
display(tabla_final_esp)
```
