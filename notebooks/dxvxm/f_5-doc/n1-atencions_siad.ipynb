## Limpieza genérica

```python
def filtrar_enteros(df: pd.DataFrame, columna: str) -> pd.DataFrame:
    """
    Dado un DataFrame y una columna, devuelve el DataFrame modificado,
    conservando solamente las filas en las que esta columna tiene un número entero (en cualquier formato).
    La columna de entrada pasará a tener formato int.
    """

    # Comprobar que el DataFrame contiene la columna
    if columna not in df.columns:
        raise ValueError(f"La columna '{columna}' no existe en el DataFrame")

    # Convertir a numérico; los valores no válidos pasan a NaN
    valores = pd.to_numeric(df[columna], errors="coerce")

    # Conservar únicamente los valores enteros
    mascara = valores.notna() & (valores % 1 == 0)

    # Filtrar las filas y convertir la columna a int
    df = df.loc[mascara].copy()
    df[columna] = valores.loc[mascara].astype(int)

    return df
```

#### dxvxm-catalunya-v_1_4-cat

```python
df_siad_mujeres_por_anyo_cat = (
    df_siad_1_anyos_enteros[df_siad_1_anyos_enteros['deteccioviolenciamasclista'] == 'Sí']  # Filtro de violencia de género
    .groupby('any')['count']                                        # Agrupar por año
    .sum()                                                          # Contar las filas
    .reindex(range(2010, 2025))                                     # Añadir años vacíos
    .reset_index()                                                  # Reestandarizar índice después de agrupar
)

# Renombrar columnas
df_siad_mujeres_por_anyo_cat = df_siad_mujeres_por_anyo_cat.rename(columns={
    'any': 'Any',
    'count': 'Dones ateses per VdG'
})

# Usar nullable integer type (para que los años sin datos aparezcan como <NA>)
df_siad_mujeres_por_anyo_cat['Dones ateses per VdG'] = (
    df_siad_mujeres_por_anyo_cat['Dones ateses per VdG'].astype('Int64')
)

df_siad_mujeres_por_anyo_cat.head(15)
```

#### dxvxm-generico-v_1_5_4

```python
# Necesitamos un mapeo de comarcas a provincias

# Podemos probar con el dataset de divisiones administrativas del ICGC (https://www.icgc.cat/ca/Geoinformacio-i-mapes/Dades-i-productes/Geoinformacio-cartografica/Divisions-administratives)
    # En este caso, se ha descargado como CSV

df_divisions = pd.read_csv(PATH_DATA / "divisions-administratives-v2r2-caps-municipi-20260120.csv", sep=";")
df_divisions.head(10)
```

```python
# Primero vamos a mirar qué comarcas hay diferentes entre cada DataFrame

missing = df_siad_1_anyos_enteros.loc[
    ~df_siad_1_anyos_enteros['comarca'].isin(df_divisions['Comarca']),
    'comarca'
]
print("Comarcas del dataset de los SIAD que faltan en las divisiones:\n", missing.unique())
print("\n")

missing = df_divisions.loc[
    ~df_divisions['Comarca'].isin(df_siad_1_anyos_enteros['comarca']),
    'Comarca'
]
print("Comarcas del dataset de divisiones que faltan en el de los SIAD:\n", missing.unique())
```

'Conselh Generau d'Aran' se puede mapear a 'Val d'Aran', pero hay que verificar que no haya ningún SIAD en el Lluçanès.

La forma correcta es mirando qué poblaciones hay en el Lluçanès y asegurando que ningún SIAD se ubica en estas poblaciones.

En el [dataset directorio de los SIADs](https://analisi.transparenciacatalunya.cat/Societat-benestar/Directori-dels-serveis-d-informaci-i-atenci-a-les-/fqcc-7vme/about_data), intentamos filtrar por POBLACIÓ. Buscamos estos municipios. en el buscador del filtro. Resultado: no se encuentra ninguno de ellos, con lo que se asume que el Lluçanès es la única comarca sin centros del SIAD.

```python
df_siad_1_3 = df_siad_1_anyos_enteros.copy()
df_siad_1_3['comarca'] = (df_siad_1_anyos_enteros['comarca'].replace("Conselh Generau d'Aran", "Val d'Aran"))
```

```python
all(df_siad_1_3['comarca'].isin(df_divisions['Comarca']))
```

```python
# Para el merge, necesitamos un dataset de mapeo con una fila por comarca

df_comarcas_provincias = (
    df_divisions[['Comarca', 'Província']]
    .drop_duplicates(subset='Comarca')
)

df_comarcas_provincias.head()
```

```python
# Añadimos la provincia en base al dataset de divisiones

df_siad_1_4 = df_siad_1_3.merge(
    df_comarcas_provincias,
    left_on='comarca',
    right_on='Comarca',
    how='left',
    validate='many_to_one'
)
df_siad_1_4 = df_siad_1_4.drop(columns='Comarca')
df_siad_1_4 = df_siad_1_4.rename(columns={'Província': 'provincia'})

print(df_siad_1_4.info())
df_siad_1_4.head(20)
```

```python
# Total women attended per province
df_siad_mujeres_provincias_cat = df_siad_1_4.groupby('provincia').agg(
    **{
        'Total dones ateses': ('count', 'sum'),
        'Dones ateses per VM': (
            'count',
            lambda x: x[df_siad_1_4.loc[x.index, 'deteccioviolenciamasclista'] == 'Sí'].sum()
        )
    }
)

# Calculate percentage
df_siad_mujeres_provincias_cat['% dones ateses per VM'] = (
    df_siad_mujeres_provincias_cat['Dones ateses per VM'] /
    df_siad_mujeres_provincias_cat['Total dones ateses'] * 100
).round(2)

# Format percentage
df_siad_mujeres_provincias_cat['% dones ateses per VM'] = (
    df_siad_mujeres_provincias_cat['% dones ateses per VM'].map(lambda x: f'{x:.2f}'.replace('.', ',') + '%')
)

# Reorder columns
df_siad_mujeres_provincias_cat = df_siad_mujeres_provincias_cat[
    ['% dones ateses per VM', 'Total dones ateses', 'Dones ateses per VM']
]

# Order provinces
df_siad_mujeres_provincias_cat = df_siad_mujeres_provincias_cat.reindex(
    ['Girona', 'Tarragona', 'Lleida', 'Barcelona']
)

df_siad_mujeres_provincias_cat.head()
```

```python
# Renombrar columnas
df_siad_mujeres_provincias_esp = df_siad_mujeres_provincias_cat.rename(
    columns={
        "% dones ateses per VM": "% mujeres atendidas por VM",
        "Total dones ateses": "Total mujeres atendidas",
        "Dones ateses per VM": "Mujeres atendidas por VM",
    }
)
df_siad_mujeres_provincias_esp.head()
```
