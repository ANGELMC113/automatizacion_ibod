## Descarga manual

Instrucciones para generar el dataset:

- Acceder a: https://estadisticasviolenciagenero.igualdad.gob.es/

- Ámbito de análisis: 010 Feminicidios en la pareja o expareja

- Columnas: Feminicidios pareja o expareja, Huérfanas y huérfanos menores de edad -1-

- Filas: Comunidad autónoma (As), Provincia (As), Año, Denuncia -2-, 

- Descargar como CSV y llamarlo d1-feminicidios_raw.csv (PATH_DATASET_1)


Notas de la fuente:

- -1-: Solo se contabiliza información relacionada con este indicador a partir de 2013.

- -2-: Denuncias: todos los asuntos que se trasladan al juzgado (incluye parte de lesiones y actuación policial de oficio).


Notas del análisis:

- El archivo analizado se corresponde con la fecha de actualización 30/10/2026

- El año más reciente es el actual. Contiene datos, pero están incompletos.

```python
d1_raw = pd.read_csv(PATH_DATA / "d1-feminicidios_raw.csv")
d1_raw.head(20)
```

```python
d1_raw.dtypes
```

Con el formato indicado para la descarga, tenemos columnas de 2 tipos:

- Columnas de variables:
    - Comunidad autónoma (As)
    - Provincia (As)
    - Año
    - Denuncia -2-

- Columna de indicador:
    - Feminicidios pareja o expareja
    - Huérfanas y huérfanos menores de edad -1-

Este formato facilita:

- Filtrar según las variables fácilmente.
- Coger la medida objetivo fácilmente.

Además, cuando ambos indicadores son 0, la fila desaparece, lo que reduce el número de valores (con esto hay que tener cuidado).

```python
# Por ejemplo:
    # ¿Cuántos feminicidios (indicador) hubo en la Provincia de Barcelona (filtro 1) en 2023 (filtro 2)?

total = d1_raw.loc[
    (d1_raw["Provincia (As)"] == "Barcelona") &             # Filtro 1
    (d1_raw["Año"] == 2023),                                # Filtro 2
    "Feminicidios pareja o expareja"                        # Indicador
].sum()

print(f"Feminicidios en Barcelona en 2023: {int(total)}")
```

## Asset 1: pasar valores a enteros y añadir agregaciones

```python
# Comprobamos que todos los números son enteros
columnas_numericas = d1_raw.select_dtypes(include='number').columns

assert (d1_raw[columnas_numericas] % 1 == 0).all().all(), "Error de formato. Un valor que debía ser un entero no se ha podido convertir a entero."

# Pasamos los valores de las columnas numéricas a enteros
columnas_numericas = d1_raw.select_dtypes(include='number').columns
d1_raw[columnas_numericas] = d1_raw[columnas_numericas].astype(int)

d1_raw.dtypes
```
